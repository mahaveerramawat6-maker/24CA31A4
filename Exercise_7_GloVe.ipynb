# Exercise 7: Word Embeddings using GloVe

## Aim
To understand the concept of GloVe (Global Vectors for Word Representation) and use pre-trained GloVe embeddings to measure the semantic similarity between two sentences.

## Objectives
After completing this experiment, students will be able to:
1. Understand the concept of word embeddings.
2. Understand the working principle of GloVe.
3. Explain how GloVe uses global word co-occurrence statistics.


## Pre-Lab
Given: “King and Queen rule the kingdom”; “Prince belongs to the royal family”; “Cat drinks milk”; “Dog drinks milk”.

**1. Similar words:** King, Queen and Prince are related to royalty; Cat and Dog are animals.  
**2. King and Queen:** they refer to related royal roles and appear in similar contexts.  
**3. How GloVe learns:** it uses global word co-occurrence statistics from a large corpus.  
**4. Benefit of pre-trained embeddings:** they provide vectors already learned from large amounts of text.  
**5. Two applications:** text classification and semantic similarity.


## In-Lab
Algorithm: import libraries, load pre-trained GloVe vectors, define two sentences, look up word vectors, average them, compute cosine similarity.

**Note:** You must download a GloVe text vector file separately and set `GLOVE_FILE` to its location. The pretrained vectors are not embedded in this notebook.


```python
import os
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

GLOVE_FILE = "glove.6B.50d.txt"  # Change this to the downloaded file path

def load_glove(path):
    vectors = {}
    with open(path, "r", encoding="utf-8") as handle:
        for line in handle:
            parts = line.rstrip().split()
            if len(parts) < 3:
                continue
            try:
                vectors[parts[0]] = np.asarray(parts[1:], dtype=np.float32)
            except ValueError:
                continue
    return vectors

if os.path.isfile(GLOVE_FILE):
    glove = load_glove(GLOVE_FILE)
    print("Loaded word vectors:", len(glove))
    print("Vector dimension:", len(next(iter(glove.values()))) if glove else 0)
else:
    glove = {}
    print("GloVe file not found. Download a GloVe text file and update GLOVE_FILE.")
```

```python
def sentence_embedding(sentence, embeddings):
    found = []
    for word in sentence.lower().split():
        word = word.strip(".,!?;:\"'()[]{}")
        if word in embeddings:
            found.append(embeddings[word])
    return np.mean(found, axis=0) if found else None

sentence1 = "I love machine learning"
sentence2 = "I enjoy artificial intelligence"

if glove:
    v1 = sentence_embedding(sentence1, glove)
    v2 = sentence_embedding(sentence2, glove)
    if v1 is not None and v2 is not None:
        score = cosine_similarity(v1.reshape(1,-1), v2.reshape(1,-1))[0,0]
        print("Sentence 1:", sentence1)
        print("Sentence 2:", sentence2)
        print("Cosine Similarity:", round(float(score), 4))
    else:
        print("At least one sentence has no known words in the loaded vectors.")
else:
    print("Similarity not computed because GloVe vectors are not loaded.")
```

## Result
The notebook loads pre-trained GloVe vectors, averages available word vectors into sentence vectors and calculates cosine similarity once the external vector file is provided.

## Post-Lab
**1. What is GloVe?** A method that learns word vectors using global co-occurrence statistics.  
**2. Global Vectors:** numerical representations learned from corpus-wide word co-occurrence information.  
**3. Difference from Word2Vec:** GloVe uses global co-occurrence counts; Word2Vec predicts words from local context.  
**4. Why pre-trained?** The vectors have already been learned from a large corpus before use.
