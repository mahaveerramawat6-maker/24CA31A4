# Exercise 10: Fine-tune a BERT-based Model for Named Entity Recognition (NER)

## Aim
To fine-tune a pretrained BERT model for Named Entity Recognition (NER) and identify entities such as Person, Organization, Location, Date, and Money from text.

## Objectives
After completing this experiment, students will be able to:
1. Understand the concept of pretrained BERT models.
2. Explain the process of fine-tuning BERT for Named Entity Recognition.


## Pre-Lab

Given:
- S1: Sundar Pichai is the CEO of Google.
- S2: Microsoft was founded by Bill Gates.
- S3: The conference will be held in Chennai on 20 July 2026.
- S4: Elon Musk visited India last week.

**1. Person names:** Sundar Pichai, Bill Gates, Elon Musk.  
**2. Organizations:** Google, Microsoft.  
**3. Locations:** Chennai, India.  
**4. Date:** 20 July 2026. “Last week” is a relative time expression.  
**5. NER:** a technique that locates named entities in text and classifies them into types.  
**6. Pretrained BERT:** a transformer model trained on a large text corpus before being adapted to a task.  
**7. Why fine-tune BERT?** Fine-tuning adapts pretrained representations to labelled examples for a particular task such as NER.


## In-Lab

### Workflow
Input sentence → load pretrained token-classification model → tokenize → predict entity spans → display entities.

**Implementation note:** The following cell uses a pretrained BERT-family NER checkpoint for inference. Full fine-tuning needs a labelled NER dataset, matching entity labels, and training compute. This distinction is made explicit rather than claiming that inference is fine-tuning.


```python
# If required, install in the active notebook environment:
# %pip install transformers torch

from transformers import pipeline

# General-purpose pretrained BERT-family NER model.
# First run downloads model files and requires internet access.
ner_pipeline = pipeline(
    "token-classification",
    model="dslim/bert-base-NER",
    aggregation_strategy="simple"
)
print("Pretrained NER pipeline loaded.")
```

```python
sentences = [
    "Sundar Pichai is the CEO of Google.",
    "Microsoft was founded by Bill Gates.",
    "The conference will be held in Chennai on 20 July 2026.",
    "Elon Musk visited India last week."
]

for sentence in sentences:
    print("\nSentence:", sentence)
    entities = ner_pipeline(sentence)
    if not entities:
        print("No entities detected.")
    for entity in entities:
        print(f"{entity['word']} -> {entity['entity_group']} (confidence: {entity['score']:.3f})")
```

## Observation
The model predicts entity spans and confidence scores. A general-purpose checkpoint may not identify every date, money value or entity type exactly as expected. For actual custom fine-tuning, prepare labelled examples and train with labels supported by the selected model/tokenizer.


## Result
A pretrained BERT-based NER pipeline was loaded and used to identify named entities in the provided sentences. The detected spans, labels and confidence scores were displayed.
