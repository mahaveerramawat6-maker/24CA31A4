### Pre-Lab

#### 1. Define tokenization with an example.

Tokenization is the process of dividing a sentence or text into smaller meaningful units called tokens. These tokens can be words, characters, or subwords, depending on the application. It is one of the first and most important preprocessing steps in Natural Language Processing (NLP).
Example: The sentence "I love Artificial Intelligence" is tokenized into ["I", "love", "Artificial", "Intelligence"].
Tokenization helps computers understand, analyze, and process text for tasks such as text classification, sentiment analysis, machine translation, and chatbots.

#### 2. What are stop words? Give five examples.

Stop words are commonly used words in a language that usually do not add significant meaning to a sentence and are often removed during text preprocessing in Natural Language Processing (NLP) to improve efficiency.

Examples of stop words: the, is, and, in, of.

Removing stop words helps reduce unnecessary data and improves the performance of text analysis and machine learning models.

#### 3. Define stemming. Give one example.

Stemming is the process of reducing a word to its root or base form by removing prefixes or suffixes. It is a common preprocessing technique in Natural Language Processing (NLP) that helps group similar words together.

Example: playing, played, and plays are reduced to the stem play.

Stemming improves text analysis, search engines, and machine learning by treating related words as the same root word.

#### 4. What is lemmatization? How does it differ from stemming?

Lemmatization is the process of converting a word into its base or dictionary form (lemma) using vocabulary and grammar rules. It produces meaningful words and is widely used in Natural Language Processing (NLP).

Example: running → run, better → good.

Difference from Stemming: Stemming removes prefixes or suffixes without considering the word's meaning, so it may produce incorrect roots. Lemmatization uses linguistic rules and a dictionary to produce meaningful base words, making it more accurate than stemming.

#### 5. What is Part-of-Speech (POS) tagging?

Part-of-Speech (POS) tagging is the process of assigning a grammatical category to each word in a sentence, such as noun, verb, adjective, adverb, or pronoun. It helps computers understand the role of each word in a sentence.

Example: In the sentence "The cat runs fast", The (Determiner), cat (Noun), runs (Verb), and fast (Adverb).

POS tagging is an important NLP technique used in text analysis, machine translation, information extraction, and sentiment analysis.

#### 6. What is Named Entity Recognition (NER)? for this pre-lab i want the answer for each question elaborately in visual learning in landscape

Named Entity Recognition (NER) is a Natural Language Processing (NLP) technique used to identify and classify important entities in a text.
It recognizes entities such as Person, Organization, Location, Date, Time, and Money.
For example, in "Ravi works at Microsoft in Hyderabad", Ravi is a Person, Microsoft is an Organization, and Hyderabad is a Location.
NER helps computers understand the meaning and context of text more accurately.
It is widely used in chatbots, search engines, information extraction, and question-answering systems.

```python
import nltk
import string

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
text = "Dr. Ravi is studying Machine Learning at Anna University in Chennai."
```

```python
tokens = word_tokenize(text)
```

```python
stop_words = set(stopwords.words("english"))

filtered_tokens = [
    word.replace(".", "")
    for word in tokens
    if word.lower() not in stop_words
    and word not in string.punctuation
]
```

#### STEP 3-LEMMATIZER

```python
lemmatizer = WordNetLemmatizer()

lemmatized_tokens = [
    lemmatizer.lemmatize(word)
    for word in filtered_tokens
]
```

```python
print("Filtered Tokens:")
print(filtered_tokens)

print("\nLemmatized Tokens:")
print(lemmatized_tokens)
```
