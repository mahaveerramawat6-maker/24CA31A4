# Exercise 4: TF-IDF (Term Frequency–Inverse Document Frequency)

## Aim

To implement TF-IDF for converting textual data into weighted numerical features and identifying important words in a document collection.

## Objective

Create vocabulary, calculate TF, DF and IDF, generate the TF-IDF matrix and interpret the resulting weights.

## Pre-Lab Task:

### 1. Define Term Frequency (TF).

TF measures how often a term occurs in a document, commonly divided by the document's total token count.

### 2. Define Document Frequency (DF).

DF is the number of documents in a corpus that contain a term.

### 3. What is IDF?

Inverse Document Frequency gives lower weights to terms found in many documents and higher weights to terms found in fewer documents.

### 4. Write the TF-IDF formula.

TF-IDF(t,d) = TF(t,d) × IDF(t). A common smoothed IDF formula is ln((1+N)/(1+DF(t))) + 1.

### 5. Why is TF-IDF useful?

It emphasizes terms that are frequent in one document but less common across the full collection.

## In- Lab Task

### Tasks

1. Create a themed document collection.
2. Create a vocabulary and TF matrix.
3. Calculate DF and smoothed IDF.
4. Calculate and display TF-IDF weights.
5. Interpret the terms with larger weights.

#### 1. Create a different document corpus.

```python
documents = [
    "solar panels convert sunlight into electricity",
    "wind turbines generate clean electricity",
    "sunlight supports solar energy production"
]
doc_names = ["D1", "D2", "D3"]
tokens = [d.lower().split() for d in documents]
vocabulary = sorted(set(w for d in tokens for w in d))
print("Vocabulary:", vocabulary)
```

#### 2. Calculate Term Frequency.

```python
import pandas as pd
from collections import Counter
tf = pd.DataFrame(
    [{w: Counter(row)[w] / len(row) for w in vocabulary} for row in tokens],
    index=doc_names
).fillna(0)
display(tf.round(4))
```

#### 3. Calculate Document Frequency and IDF.

```python
import numpy as np
N = len(documents)
df = {w: sum(w in set(row) for row in tokens) for w in vocabulary}
idf = {w: np.log((1 + N) / (1 + df[w])) + 1 for w in vocabulary}
display(pd.DataFrame({"DF": df, "IDF": idf}).round(4))
```

#### 4. Construct the TF-IDF matrix.

```python
tfidf = tf.copy()
for word in vocabulary:
    tfidf[word] = tf[word] * idf[word]
display(tfidf.round(4))
```

#### 5. Display the highest TF-IDF value in each document.

```python
for name, row in tfidf.iterrows():
    print(name, "highest-weight term:", row.idxmax(), "score:", round(row.max(), 4))
```

## Post Lab

### Live Scenario: Comparing renewable energy descriptions

A research group compares short descriptions of solar, wind and hydroelectric power. Use a new dataset to calculate TF-IDF and identify terms that distinguish the documents.

#### Sample Dataset

```python
post_docs = [
    "hydroelectric dams use flowing water",
    "solar cells use sunlight",
    "flowing water can produce electricity"
]
ptokens = [d.lower().split() for d in post_docs]
pvocab = sorted(set(w for d in ptokens for w in d))
ptf = pd.DataFrame([{w: Counter(row)[w]/len(row) for w in pvocab} for row in ptokens],
                   index=["D1","D2","D3"]).fillna(0)
pdf = {w:sum(w in set(row) for row in ptokens) for w in pvocab}
pidf = {w:np.log((1+len(ptokens))/(1+pdf[w]))+1 for w in pvocab}
ptfidf = ptf.copy()
for w in pvocab: ptfidf[w] = ptf[w]*pidf[w]
display(ptfidf.round(4))
```
