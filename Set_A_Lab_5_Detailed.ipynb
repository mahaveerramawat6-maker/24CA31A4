# Exercise 5: Text Classification using Naïve Bayes and SVM

## Aim

To implement text classification using Naïve Bayes for spam detection and Support Vector Machine for sentiment analysis.

## Objective

Convert text to TF-IDF features, train classifiers, evaluate predictions and classify new examples.

## Pre-Lab Task:

### 1. What is text classification?

Text classification assigns a predefined label to a text, such as Spam/Not Spam or Positive/Negative.

### 2. What is Naïve Bayes?

Naïve Bayes is a probabilistic classifier based on Bayes' theorem with a simplifying conditional-independence assumption.

### 3. What is SVM?

Support Vector Machine finds a decision boundary that separates classes in a feature space.

### 4. Why use TF-IDF?

TF-IDF converts text into numerical features and can emphasize informative terms.

### 5. What does accuracy measure?

Accuracy is the proportion of predictions that match the true labels.

## In- Lab Task

### Tasks

1. Prepare labelled messages.
2. Convert messages to TF-IDF features.
3. Split data into training and test sets.
4. Train Naïve Bayes and evaluate spam labels.
5. Train SVM for sentiment analysis.
6. Predict classes for new text.

#### 1. Import libraries and create spam-message data.

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

messages = [
    "You are today's lucky winner", "Claim your cash bonus",
    "Team practice begins at six", "Please review the meeting notes",
    "Exclusive reward waiting for you", "Submit the lab report",
    "You won a shopping voucher", "The lecture starts on Monday",
    "Claim bonus points now", "Your timetable has been updated",
    "Get a free gift today", "Bring your notebook to class"
]
labels = ["Spam","Spam","Not Spam","Not Spam","Spam","Not Spam",
          "Spam","Not Spam","Spam","Not Spam","Spam","Not Spam"]
print("Messages:", len(messages))
```

#### 2. Split data and create TF-IDF features.

```python
X_train, X_test, y_train, y_test = train_test_split(
    messages, labels, test_size=0.25, random_state=17, stratify=labels
)
spam_vectorizer = TfidfVectorizer()
X_train_vec = spam_vectorizer.fit_transform(X_train)
X_test_vec = spam_vectorizer.transform(X_test)
print("Training matrix:", X_train_vec.shape)
```

#### 3. Train and evaluate Naïve Bayes.

```python
nb_model = MultinomialNB().fit(X_train_vec, y_train)
spam_pred = nb_model.predict(X_test_vec)
print("Accuracy:", round(accuracy_score(y_test, spam_pred), 3))
print(confusion_matrix(y_test, spam_pred, labels=["Not Spam","Spam"]))
print(classification_report(y_test, spam_pred, zero_division=0))
```

#### 4. Predict spam labels for new messages.

```python
new_messages = ["Claim a free shopping voucher", "Please attend the lab session"]
for text, label in zip(new_messages, nb_model.predict(spam_vectorizer.transform(new_messages))):
    print(text, "->", label)
```

#### 5. Prepare review data for SVM sentiment analysis.

```python
reviews = [
    "The planetarium show was fascinating", "The exhibition was disappointing",
    "The visit was average", "Wonderful guide and displays",
    "Boring and poorly arranged", "It was an ordinary afternoon",
    "I enjoyed every gallery", "The ticket service was awful",
    "The place was acceptable", "Beautiful collection",
    "Waste of money", "Nothing special"
]
sentiments = ["Positive","Negative","Neutral"] * 4
print("Reviews:", len(reviews))
```

#### 6. Train and evaluate SVM.

```python
r_train, r_test, s_train, s_test = train_test_split(
    reviews, sentiments, test_size=0.25, random_state=9, stratify=sentiments
)
review_vectorizer = TfidfVectorizer()
r_train_vec = review_vectorizer.fit_transform(r_train)
r_test_vec = review_vectorizer.transform(r_test)
svm_model = LinearSVC().fit(r_train_vec, s_train)
sent_pred = svm_model.predict(r_test_vec)
print("Accuracy:", round(accuracy_score(s_test, sent_pred), 3))
print(confusion_matrix(s_test, sent_pred, labels=["Negative","Neutral","Positive"]))
```

## Post Lab

### Live Scenario: Categorizing science, sport and culture news

A campus bulletin board wants to sort short news headlines into Sports, Science and Culture categories. Train a simple text classifier on labelled sample headlines.

#### Sample Dataset

```python
news = [
    "The hockey team won the final", "A player scored a goal",
    "Researchers tested a solar battery", "Scientists studied ocean samples",
    "The orchestra performed a new symphony", "The film festival opened",
    "The coach announced the team lineup", "A laboratory published findings",
    "The director received an award"
]
categories = ["Sports","Sports","Science","Science","Culture","Culture","Sports","Science","Culture"]
news_vectorizer = TfidfVectorizer()
news_matrix = news_vectorizer.fit_transform(news)
news_model = LinearSVC().fit(news_matrix, categories)
samples = ["The hockey team won", "Scientists tested a battery", "The orchestra performed"]
for text, label in zip(samples, news_model.predict(news_vectorizer.transform(samples))):
    print(text, "->", label)
```
