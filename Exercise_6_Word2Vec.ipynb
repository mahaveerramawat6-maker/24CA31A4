## Pre-Lab
Corpus: “I love machine learning”; “I enjoy artificial intelligence”; “Cats drink milk”; “Dogs drink milk”.

**1. Similar words:** love/enjoy; cats/dogs are related animal terms.  
**2. Numerical representation:** Word2Vec learns dense vectors from surrounding context.  
**3. Predicts the missing/target word:** CBOW.  
**4. Predicts surrounding words:** Skip-Gram.


## In-Lab
Algorithm: import libraries, create and tokenize corpus, train Word2Vec, create word vectors, average vectors for sentence embeddings, calculate cosine similarity.

```python
# If required: %pip install gensim scipy scikit-learn
import re
import numpy as np
from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity

sentences = [
 "I love machine learning", "I enjoy artificial intelligence",
 "Cats drink milk", "Dogs drink milk", "Machine learning uses data",
 "Artificial intelligence learns patterns", "Dogs and cats are animals",
 "I enjoy learning new technology"
]
tokens = [re.findall(r"\b\w+\b", s.lower()) for s in sentences]
print("Tokenized corpus:")
for row in tokens: print(row)
```

```python
model = Word2Vec(sentences=tokens, vector_size=50, window=3, min_count=1,
                  workers=1, sg=1, seed=42, epochs=100)
print("Model trained:", bool(model.wv.key_to_index))
print("Vocabulary:", list(model.wv.key_to_index))
```

```python
def sentence_vector(text):
    words = re.findall(r"\b\w+\b", text.lower())
    vectors = [model.wv[w] for w in words if w in model.wv]
    return np.mean(vectors, axis=0) if vectors else np.zeros(model.vector_size)

s1, s2 = "I love machine learning", "I enjoy artificial intelligence"
v1, v2 = sentence_vector(s1), sentence_vector(s2)
score = cosine_similarity(v1.reshape(1,-1), v2.reshape(1,-1))[0,0]
print("Sentence 1:", s1)
print("Sentence 2:", s2)
print("Cosine Similarity:", round(float(score), 4))
```

## Result
Word2Vec embeddings were trained on the sample corpus and cosine similarity was computed between sentence vectors. Values depend on the small corpus and random initialization.

## Post-Lab
**1. Word2Vec:** a method that learns dense word vectors from context.  
**2. Advantage over BoW:** embeddings can represent semantic relationships; BoW mainly records counts.  
**3. CBOW vs Skip-Gram:** CBOW predicts a word from context, while Skip-Gram predicts context from a word.
