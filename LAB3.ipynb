### 6. What is Named Entity Recognition (NER)?

**Named Entity Recognition (NER)** is an NLP technique used to identify and classify important named entities in text.

Common entity types include:

- **PERSON** – names of people
- **ORGANIZATION** – companies, institutions, or groups
- **LOCATION** – cities, states, countries, and places
- **DATE** – dates and years
- **MONEY** – monetary values
- **TIME** – time expressions

**Example:**

> "Ravi studies at Anna University in Chennai."

A Named Entity Recognition system can identify:

- **Ravi → PERSON**
- **Anna University → ORGANIZATION**
- **Chennai → LOCATION**

NER is useful in information extraction, search engines, chatbots, question answering, document analysis, and many other NLP applications.

## Activity 1: One-Hot Encoding

### Tasks

1. Import the required Python libraries.
2. Create a vocabulary from the given corpus.
3. Identify all unique words.
4. Generate One-Hot Encoding vectors for each unique word.
5. Display the One-Hot Encoding vectors in tabular format.


```python
# Import required libraries
import pandas as pd
import numpy as np

# Given corpus
documents = [
    "I love NLP",
    "I love Python",
    "NLP is fun"
]

document_names = ["D1", "D2", "D3"]

print("Corpus:")
for name, doc in zip(document_names, documents):
    print(f"{name}: {doc}")
```

```python
# Create the vocabulary from the corpus

# Convert all words to lowercase
all_words = []
for document in documents:
    all_words.extend(document.lower().split())

# Identify unique words and arrange them alphabetically
vocabulary = sorted(set(all_words))

print("Unique Words (Vocabulary):")
print(vocabulary)

print("\nVocabulary Size:", len(vocabulary))
```

```python
# Generate One-Hot Encoding vectors

one_hot_vectors = {}

for word in vocabulary:
    vector = [1 if vocabulary[i] == word else 0 for i in range(len(vocabulary))]
    one_hot_vectors[word] = vector

one_hot_df = pd.DataFrame(
    one_hot_vectors,
    index=vocabulary
).T

one_hot_df.columns = vocabulary

print("One-Hot Encoding Vectors:")
display(one_hot_df)
```

### One-Hot Encoding Observation

Each unique word is represented by a vector whose length equals the vocabulary size.

For this corpus:

**Vocabulary size = 6**

Therefore, every One-Hot Encoding vector contains **6 positions** and exactly one position contains **1**.


# Activity 2: Bag of Words (BoW)

### Tasks

1. Create the Bag of Words model using `CountVectorizer`.
2. Generate the vocabulary.
3. Compute the frequency of each word in every document.
4. Construct the Document-Term Matrix (DTM).
5. Display the BoW matrix.


```python
# Import CountVectorizer
from sklearn.feature_extraction.text import CountVectorizer

# Create the Bag of Words model
vectorizer = CountVectorizer()

# Fit the model and transform the documents
bow_matrix = vectorizer.fit_transform(documents)

print("Bag of Words model created successfully.")
```

```python
# Generate the vocabulary

bow_vocabulary = vectorizer.get_feature_names_out()

print("BoW Vocabulary:")
print(list(bow_vocabulary))
```

```python
# Construct the Document-Term Matrix (DTM)

dtm = pd.DataFrame(
    bow_matrix.toarray(),
    index=document_names,
    columns=bow_vocabulary
)

print("Document-Term Matrix (DTM):")
display(dtm)
```

### Understanding the DTM

The **Document-Term Matrix (DTM)** represents:

- **Rows → Documents**
- **Columns → Words**
- **Cell values → Number of times a word occurs in that document**

For example, the word **love** occurs once in D1, once in D2, and zero times in D3.


```python
# Display word frequencies in every document

print("Word Frequencies:")
display(dtm)

print("\nFrequency of 'love' in each document:")
print(dtm["love"])
```

# Activity 3: Word Cloud

### Tasks

1. Combine all documents into a single text.
2. Generate a Word Cloud.
3. Display the Word Cloud.
4. Identify the top five most frequent words from the visualization.


```python
# Import WordCloud and Matplotlib
import matplotlib.pyplot as plt
from wordcloud import WordCloud
```

```python
# Combine all documents into a single text

combined_text = " ".join(documents)

print("Combined Corpus:")
print(combined_text)
```

```python
# Generate and display the Word Cloud

wordcloud = WordCloud(
    width=1200,
    height=600,
    background_color="white"
).generate(combined_text.lower())

plt.figure(figsize=(14, 7))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title("Word Cloud of the Given Corpus", fontsize=18)
plt.show()
```

```python
# Identify the top five most frequent words

word_frequencies = wordcloud.words_

# Convert normalized WordCloud frequencies into a ranked table
frequency_table = (
    pd.DataFrame(
        list(word_frequencies.items()),
        columns=["Word", "Normalized Frequency"]
    )
)

# Add actual corpus counts
actual_counts = (
    pd.Series(combined_text.lower().split())
    .value_counts()
    .rename_axis("Word")
    .reset_index(name="Count")
)

top_five = actual_counts.head(5)

print("Top Five Most Frequent Words:")
display(top_five)
```
