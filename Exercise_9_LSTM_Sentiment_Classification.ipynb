# Exercise 9: Build an LSTM Model for Sentiment Classification

## Aim
To build and evaluate a Long Short-Term Memory (LSTM) model for classifying text sentiment into Positive and Negative classes using deep learning techniques.

## Objectives
1. Understand Recurrent Neural Networks (RNN) and Long Short-Term Memory (LSTM).
2. Perform text preprocessing using tokenization and padding.
3. Convert text into numerical sequences suitable for deep learning models.
4. Build an LSTM-based sentiment classification model using TensorFlow/Keras.
5. Train and evaluate the LSTM model.
6. Predict the sentiment of unseen text.


## Pre-Lab
Reviews:
- The product quality is excellent.
- I am very happy with the service.
- The delivery was very slow.
- The product is damaged.
- Excellent customer support.
- Very disappointed with the purchase.

**1. Sentiment labels:** Positive, Positive, Negative, Negative, Positive, Negative, respectively.  
**2. Train an LSTM:** tokenize the labelled reviews, pad sequences, build an Embedding–LSTM–Dense model and train it. This tiny dataset is for demonstration only.  
**3. “The product is very good and easy to use.”** Expected sentiment: Positive, because “very good” expresses favourable opinion.


## In-Lab
Architecture: Embedding maps token IDs to vectors; LSTM learns sequence patterns; Dense produces a prediction; Sigmoid outputs a probability between 0 and 1.

```python
# If needed: %pip install tensorflow
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout

tf.keras.utils.set_random_seed(42)
reviews = [
 "The product quality is excellent",
 "I am very happy with the service",
 "The delivery was very slow",
 "The product is damaged",
 "Excellent customer support",
 "Very disappointed with the purchase"
]
labels = np.array([1, 1, 0, 0, 1, 0], dtype=np.float32)  # 1=Positive, 0=Negative
tokenizer = Tokenizer(oov_token="<OOV>")
tokenizer.fit_on_texts(reviews)
sequences = tokenizer.texts_to_sequences(reviews)
max_len = max(len(seq) for seq in sequences)
X = pad_sequences(sequences, maxlen=max_len, padding="post")
print("Vocabulary size:", len(tokenizer.word_index)+1)
print("Padded data shape:", X.shape)
```

```python
model = Sequential([
    Embedding(input_dim=len(tokenizer.word_index)+1, output_dim=16),
    LSTM(16),
    Dropout(0.2),
    Dense(1, activation="sigmoid")
])
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
model.summary()
```

```python
history = model.fit(X, labels, epochs=25, batch_size=2, verbose=0)
loss, accuracy = model.evaluate(X, labels, verbose=0)
print("Training loss:", round(float(loss), 4))
print("Training accuracy:", round(float(accuracy), 4))
print("Only six training examples are used; this is not a reliable generalization score.")
```

```python
test_review = "The product is very good and easy to use"
test_seq = tokenizer.texts_to_sequences([test_review])
test_pad = pad_sequences(test_seq, maxlen=max_len, padding="post")
prob = float(model.predict(test_pad, verbose=0)[0][0])
print("Review:", test_review)
print("Positive probability:", round(prob, 4))
print("Predicted sentiment:", "Positive" if prob >= 0.5 else "Negative")
```

## Result
An LSTM sentiment model was built and trained, and a new review was classified. The small dataset limits reliability.

## Post-Lab
**1. How does Embedding help?** It maps token IDs to dense trainable vectors.  
**2. Role of LSTM:** it learns sequence and context patterns useful for classification.  
**3. Two advantages over a traditional RNN:** gated memory helps retain longer context and reduces vanishing-gradient problems.
