## In-Lab — Activity 1: Spam Detection using Naïve Bayes
Create labelled email data, convert to TF-IDF, split data, train Naïve Bayes, predict labels and show accuracy/confusion matrix.

```python
# If required, run: %pip install scikit-learn
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

emails = [
 "Congratulations you won a lottery", "Claim your free prize now",
 "Meeting at 10 AM tomorrow", "Project report submission is due",
 "Win a free gift today", "Please attend the team meeting",
 "You won a cash reward", "The assignment deadline is tomorrow",
 "Claim your reward now", "Discuss the project today"
]
email_labels = ["Spam","Spam","Not Spam","Not Spam","Spam","Not Spam","Spam","Not Spam","Spam","Not Spam"]
Xtr, Xte, ytr, yte = train_test_split(emails, email_labels, test_size=.3, random_state=42, stratify=email_labels)
spam_vec = TfidfVectorizer()
Xtr_vec = spam_vec.fit_transform(Xtr)
Xte_vec = spam_vec.transform(Xte)
nb = MultinomialNB().fit(Xtr_vec, ytr)
pred = nb.predict(Xte_vec)
print("Accuracy:", round(accuracy_score(yte, pred), 3))
print("Confusion matrix (Not Spam, Spam):")
print(confusion_matrix(yte, pred, labels=["Not Spam","Spam"]))
print(classification_report(yte, pred, zero_division=0))
```

```python
new_emails = ["Claim your free reward now", "Please send the meeting notes"]
for text, label in zip(new_emails, nb.predict(spam_vec.transform(new_emails))):
    print(text, "->", label)
```

## Activity 2: Sentiment Analysis using SVM
Create review data, extract TF-IDF features, train SVM, predict sentiment, and display accuracy and confusion matrix.

```python
reviews = [
 "This phone is excellent", "Battery performance is poor", "The product is okay",
 "I love this useful device", "The service was terrible", "The quality is average",
 "Amazing camera and display", "Delivery was slow and disappointing",
 "It is neither good nor bad", "Very happy with the purchase",
 "The product stopped working", "The experience was ordinary"
]
sentiments = ["Positive","Negative","Neutral","Positive","Negative","Neutral",
 "Positive","Negative","Neutral","Positive","Negative","Neutral"]
rtr, rte, strn, ste = train_test_split(reviews, sentiments, test_size=.25, random_state=7, stratify=sentiments)
sent_vec = TfidfVectorizer()
rtr_vec = sent_vec.fit_transform(rtr)
rte_vec = sent_vec.transform(rte)
svm = LinearSVC().fit(rtr_vec, strn)
spred = svm.predict(rte_vec)
print("Accuracy:", round(accuracy_score(ste, spred), 3))
print("Confusion matrix (Negative, Neutral, Positive):")
print(confusion_matrix(ste, spred, labels=["Negative","Neutral","Positive"]))
print(classification_report(ste, spred, zero_division=0))
```

```python
new_reviews = ["The screen is wonderful", "I am unhappy with the poor quality", "The product is average"]
for text, label in zip(new_reviews, svm.predict(sent_vec.transform(new_reviews))):
    print(text, "->", label)
```

## Result
Naïve Bayes was used for spam detection and SVM for sentiment classification. Accuracy, confusion matrices and predictions for new examples were displayed. The small datasets are for demonstration only.

## Post-Lab: News Article Classification

```python
news = [
 "The team won the cricket final", "The striker scored two goals",
 "The government announced an education policy", "Parliament debated a tax bill",
 "The company launched a smartphone", "A new processor improves laptop performance",
 "The basketball team signed a player", "Ministers discussed the national budget",
 "A new operating system update was released"
]
categories = ["Sports","Sports","Politics","Politics","Technology","Technology","Sports","Politics","Technology"]
news_vectorizer = TfidfVectorizer()
news_matrix = news_vectorizer.fit_transform(news)
news_model = LinearSVC().fit(news_matrix, categories)
samples = ["The team won the match", "A new phone processor was announced", "The minister presented a bill"]
for text, label in zip(samples, news_model.predict(news_vectorizer.transform(samples))):
    print(text, "->", label)
```
