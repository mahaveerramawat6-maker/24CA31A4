## In-Lab — Activity 1: Data Preparation
Import libraries, create the corpus, tokenize and display the tokenized sentences.

```python
# If required: %pip install gensim scipy scikit-learn
import re
from gensim.models import Word2Vec

corpus = [
 "Machine learning is a branch of artificial intelligence",
 "Artificial intelligence helps computers learn from data",
 "Deep learning is a subset of machine learning",
 "Natural language processing uses machine learning",
 "Machine learning systems discover patterns from data",
 "Deep neural networks support artificial intelligence"
]
tokens = [re.findall(r"\b\w+\b", sentence.lower()) for sentence in corpus]
for row in tokens:
    print(row)
```

## Activity 2: Train Word2Vec (CBOW) Model
Set `sg=0` for CBOW, show the vocabulary and verify training.

```python
cbow_model = Word2Vec(
    sentences=tokens, vector_size=60, window=3, min_count=1,
    workers=1, sg=0, seed=21, epochs=120
)
print("Model trained successfully.")
print("Vocabulary:", sorted(cbow_model.wv.key_to_index))
```

## Activity 3: Generate Word Embeddings
Retrieve vectors for Machine and Learning and inspect the vector size.

```python
for word in ["machine", "learning"]:
    print("Embedding for", word, ":")
    print(cbow_model.wv[word])
    print("Dimension:", len(cbow_model.wv[word]))
```

## Activity 4: Semantic Similarity Analysis
Find the five words most similar to Learning, calculate two word-pair scores and interpret them.

```python
print("Top five words similar to 'learning':")
print(cbow_model.wv.most_similar("learning", topn=5))
for a, b in [("machine", "learning"), ("artificial", "intelligence")]:
    print(f"Similarity({a}, {b}) = {cbow_model.wv.similarity(a,b):.4f}")
```

## Result
The CBOW model was trained; vocabulary, embeddings, vector dimensions, nearest words and similarity scores were displayed. Small-corpus scores are illustrative.

## Post-Lab
Train CBOW on another relevant dataset and compare its nearest words and similarity results.

```python
health_corpus = [
 "Doctors treat patients in the hospital",
 "Nurses help doctors care for patients",
 "Hospitals use medical equipment",
 "Patients receive treatment from medical staff",
 "The nurse works in a clinic",
 "Doctors and nurses work together"
]
health_tokens = [re.findall(r"\b\w+\b", s.lower()) for s in health_corpus]
health_model = Word2Vec(sentences=health_tokens, vector_size=40, window=2,
                        min_count=1, workers=1, sg=0, seed=12, epochs=100)
print("Vocabulary:", sorted(health_model.wv.key_to_index))
print("Words similar to doctor:", health_model.wv.most_similar("doctor", topn=3))
```
