# Exercise 4: TF-IDF (Term Frequency–Inverse Document Frequency)

## Aim
To implement the TF-IDF (Term Frequency–Inverse Document Frequency) technique for converting textual data into weighted numerical features and identifying the importance of words in a document collection.

## Objective
After completing this experiment, students will be able to:
1. Create a vocabulary from a given text corpus.
2. Calculate the Term Frequency (TF) for each document.
3. Calculate the Document Frequency (DF) of each term.
4. Compute the Inverse Document Frequency (IDF) of each term.
5. Generate the TF-IDF matrix using Python.
6. Interpret the importance of words based on TF-IDF scores.


## Pre-Lab

Given: D1: Apple is sweet; D2: Apple is red; D3: Mango is sweet.

**1. Vocabulary:** apple, is, mango, red, sweet.  
**2. Vocabulary size:** 5.  
**3. Term Frequency:** TF(t,d) = count of term t in document d / total terms in d. Each document has three terms, so each appearing word has TF = 1/3.  
**4. Document Frequency:** number of documents containing a term: apple=2, is=3, mango=1, red=1, sweet=2.  
**5. Inverse Document Frequency:** using smoothed IDF, IDF(t) = ln((1+N)/(1+DF(t))) + 1, where N is the number of documents.  
**6. TF-IDF:** TF-IDF(t,d) = TF(t,d) × IDF(t). The full matrix is generated below.


## In-Lab — Activity 1: Vocabulary Creation
Tasks: import libraries, create the corpus, generate and display the vocabulary.

```python
import pandas as pd
import numpy as np
from collections import Counter

documents = ["Apple is sweet", "Apple is red", "Mango is sweet"]
doc_names = ["D1", "D2", "D3"]
tokens = [doc.lower().split() for doc in documents]
vocab = sorted(set(w for doc in tokens for w in doc))
print("Vocabulary:", vocab)
print("Vocabulary size:", len(vocab))
```

## Activity 2: TF, DF, IDF and TF-IDF Matrix
Tasks: count document terms, calculate TF, display TF, calculate DF and IDF, then display the TF-IDF matrix.

```python
tf = []
for words in tokens:
    counts = Counter(words)
    tf.append({w: counts[w] / len(words) for w in vocab})
tf_df = pd.DataFrame(tf, index=doc_names).fillna(0)
print("Term Frequency Matrix")
display(tf_df)
```

```python
N = len(documents)
df_values = {w: sum(w in set(doc) for doc in tokens) for w in vocab}
idf_values = {w: np.log((1 + N) / (1 + df_values[w])) + 1 for w in vocab}
display(pd.DataFrame({"DF": df_values, "IDF": idf_values}))
```

```python
tfidf = tf_df.copy()
for w in vocab:
    tfidf[w] = tf_df[w] * idf_values[w]
print("TF-IDF Matrix")
display(tfidf.round(4))
```

## Result
The vocabulary, TF, DF, IDF and TF-IDF matrix were generated. Terms that occur in fewer documents receive higher IDF weights.

## Post-Lab
Prepare the TF-IDF matrix for:
- D1: Python is easy
- D2: Python supports machine learning
- D3: Machine learning is powerful


```python
post_docs = ["Python is easy", "Python supports machine learning", "Machine learning is powerful"]
post_tokens = [d.lower().split() for d in post_docs]
post_vocab = sorted(set(w for d in post_tokens for w in d))
post_tf = []
for words in post_tokens:
    counts = Counter(words)
    post_tf.append({w: counts[w] / len(words) for w in post_vocab})
post_tf = pd.DataFrame(post_tf, index=["D1","D2","D3"]).fillna(0)
n = len(post_docs)
post_df = {w: sum(w in set(d) for d in post_tokens) for w in post_vocab}
post_idf = {w: np.log((1+n)/(1+post_df[w]))+1 for w in post_vocab}
post_tfidf = post_tf.copy()
for w in post_vocab:
    post_tfidf[w] = post_tf[w] * post_idf[w]
display(post_tfidf.round(4))
```
